```python
from scipy.optimize import curve_fit
def expdecay(x,pc,nu,K):
    return pc+K*(x**(-1/nu))
popt,pcov = curve_fit(expdecay,Larr[2:-2],maxxi[2:-2])
print(popt)
print(np.sqrt(np.diag(pcov)))
for j in range(2,len(Larr)-2):
    plt.errorbar(Larr[j],maxxi[j],yerr=maxxierr[j],fmt="o",color="blue")
    if j ==2:
        plt.errorbar(Larr[j],maxxi[j],yerr=maxxierr[j],fmt="o",color="blue",label="Data")
xarr = np.linspace(45,2200)

plt.plot(xarr,expdecay(xarr,*popt),label="Best Fit")
plt.hlines(popt[0],xmin=45,xmax=2156,label=fr"$p_c =$ {popt[0]:.5f} $\pm$ {np.sqrt(np.diag(pcov))[0]:.5f}",color="black")
plt.xlabel("L")
plt.ylabel(r"Occupation Probability of Maximum $\xi$, $p_{\xi, max}$")
plt.title(r"Finite Size Scaling of Maximum Correlation Length $\xi_{max}$")
plt.legend()
# plt.xscale("log")
# plt.yscale("log")
```

```python
#correlation length data collapse
from scipy.interpolate import interp1d
from scipy.optimize import curve_fit
#Larr
#Order parameter in Warr[L,p]
#pvals
pvals = np.linspace(0.2,0.8,1000)

xrange  = 1
xbase = np.linspace(-xrange,xrange,10000)
Lindices = range(2,len(Larr)-2)
def finite_size_scaling_residuals(xbase,pc,nu):
    mean = np.zeros(len(xbase))
    residuals = np.zeros(len(xbase))
    for i in Lindices:
        L = Larr[i]
        x = (pvals-pc)*L**(1/nu)
        y = Warr[i,:]/L
        f = interp1d(
            x,
            y,
            kind="linear",
            bounds_error=False,
            fill_value=np.nan,
        )
        # print(x)
        # print(y)
        regridy = f(xbase)
        mean += regridy
    mean /= len(Lindices)
    for i in Lindices:
        L = Larr[i]
        x = (pvals-pc)*L**(1/nu)
        y = Warr[i,:]/L
        f = interp1d(
            x,
            y,
            kind="linear",
            bounds_error=False,
            fill_value=np.nan,
        )
        regridy = f(xbase)
    residuals += np.square(mean-regridy)
    return residuals
p0 = [0.5926,4/3]
popt,pcov = curve_fit(finite_size_scaling_residuals,xbase,np.zeros(xbase.shape[0]),p0=p0)
pc,nu =popt
perr = np.sqrt(np.diag(pcov))
print(popt)
print(perr)
for i in Lindices:
    x = (pvals-pc)*Larr[i]**(1/nu)
    y = Warr[i,:]/(Larr[i])
    plt.plot(x,y,label=f"L={Larr[i]}")
plt.xlim(np.min(xbase),np.max(xbase))
plt.xlabel(r"$(p_i-p_c)L^{1/\nu}$")
plt.ylabel(r"$\xi_i/L$")
plt.title(r"Data Collapse of Correlation Length $\xi$")
plt.show()
plt.plot(xbase,finite_size_scaling_residuals(xbase,*popt))
plt.xlabel(r"$(p_i-p_c)L^{1/\nu}$")
plt.title("Variance of Best Fit Data Collapse From Mean")
plt.show()
```
